# 11A — Object Prompts (Standardized Copy)

**Purpose:** Creates a standardized copy of the object prompts for the threshold sweep pipeline.

```python
# Cell 1 – Configuration
import subprocess, sys, os
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "pandas", "pyarrow"], check=False)
import pandas as pd

try:
    from google.colab import drive
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB:
    import shutil
    mp = "/content/drive"
    subprocess.run(["fusermount", "-uz", mp], capture_output=True)
    if os.path.isdir(mp):
        shutil.rmtree(mp, ignore_errors=True)
    drive.mount(mp)
    DATA_DIR = "/content/drive/MyDrive/DATA/CSP-Atlas"
else:
    DATA_DIR = "/Users/piotrwilam/Data/CSP-Atlas"

INPUT_FILE = "prompts_106x50.parquet"
OUTPUT_FILE = "11A_object_prompts.parquet"

print(f"Data dir: {DATA_DIR}")
```

```python
# Cell 2 – Load and save with standardized name
input_path = os.path.join(DATA_DIR, INPUT_FILE)
output_path = os.path.join(DATA_DIR, OUTPUT_FILE)

df = pd.read_parquet(input_path)
df.to_parquet(output_path, index=False)

print(f"Loaded: {input_path}")
print(f"Saved:  {output_path}")
print(f"Shape:  {df.shape}")
print(f"Pairs:  {df.groupby(['ast_node', 'builtin_obj']).ngroups}")
print(f"Prompts per pair: {df.groupby(['ast_node', 'builtin_obj']).size().unique()}")
```

```python
# Cell 3 – Verification
df2 = pd.read_parquet(output_path)
assert len(df2) == len(df), "Row count mismatch"
assert list(df2.columns) == list(df.columns), "Column mismatch"
print(f"Verified: {len(df2)} rows, columns: {list(df2.columns)}")
print(f"\nSample:")
print(df2.head(3).to_string())
```
